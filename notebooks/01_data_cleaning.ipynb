```python
# Convert date columns to datetime format

customer_df["registration_date"] = pd.to_datetime(customer_df["registration_date"])

inventory_df["last_restock_date"] = pd.to_datetime(inventory_df["last_restock_date"])

promotions_df["start_date"] = pd.to_datetime(promotions_df["start_date"])
promotions_df["end_date"] = pd.to_datetime(promotions_df["end_date"])

sales_df["date"] = pd.to_datetime(sales_df["date"])

store_df["opening_date"] = pd.to_datetime(store_df["opening_date"])

print("Date conversion completed successfully.")
```

```python
# Save cleaned datasets

customer_df.to_csv("../data/processed/customer_master_clean.csv", index=False)

inventory_df.to_csv("../data/processed/inventory_snapshot_clean.csv", index=False)

promotions_df.to_csv("../data/processed/promotions_clean.csv", index=False)

sales_df.to_csv("../data/processed/sales_transactions_clean.csv", index=False)

sku_flags_df.to_csv("../data/processed/sku_inventory_flags_clean.csv", index=False)

sku_df.to_csv("../data/processed/sku_master_clean.csv", index=False)

store_df.to_csv("../data/processed/store_master_clean.csv", index=False)

print("All cleaned datasets saved successfully in data/processed/")
```
